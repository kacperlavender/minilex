```python
import sys
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
words = open('data/names.txt', 'r').read().splitlines()
words[:8]
```

```python
chars = sorted(list(set(''.join(words))))
stoi = {s: i + 1 for i, s in enumerate(chars)}
stoi['.'] = 0
itos = {i: s for s, i in stoi.items()}
print(itos)
```

```python
block_size = 3

# n gram
def build_dataset(words):
    x, y = [], []

    for w in words:
        context = [0] * block_size
        
        for ch in w + '.':
            ix = stoi[ch]
            x.append(context)
            y.append(ix)
            context = context[1:] + [ix]
    
    x = torch.tensor(x)
    y = torch.tensor(y)
    print(x.shape, y.shape)

    return x, y

import random

random.seed(42)
random.shuffle(words)
n1 = int(0.8*len(words)) # 0.8 trening
n2 = int(0.9*len(words)) # 0.1 walidacja

Xtr, Ytr = build_dataset(words[:n1]) # trenign
Xdev, Ydev = build_dataset(words[n1:n2]) # walidacja
Xte, Yte = build_dataset(words[n2:]) # test
```
